```python
# Import Libraries
from deep_translator import GoogleTranslator
import gradio as gr


# Translation Function
def translate_text(text, source_lang, target_lang):

    try:
        translated = GoogleTranslator(
            source=source_lang,
            target=target_lang
        ).translate(text)

        return translated

    except Exception as e:
        return f"Error: {str(e)}"


# Supported Languages
languages = [
    "english",
    "urdu",
    "french",
    "german",
    "spanish",
    "arabic",
    "hindi",
    "chinese (simplified)",
"chinese (traditional)",
    "turkish",
    "japanese",
    "italian",
    "russian"
]

custom_css = """

body {
    background: linear-gradient(to right, #eef2ff, #f8fafc);
}

.gradio-container {
    max-width: 950px !important;
    margin: auto;
    padding-top: 20px;
    font-family: 'Segoe UI', sans-serif;
}

.main-title {
    text-align: center;
    font-size: 38px;
    font-weight: bold;
    color: #111827;
    margin-bottom: 10px;
}

.sub-title {
    text-align: center;
    font-size: 18px;
    color: #4b5563;
    margin-bottom: 30px;
}

.feature-box {
    background: white;
    padding: 25px;
    border-radius: 18px;
    box-shadow: 0px 4px 20px rgba(0,0,0,0.08);
}

textarea {
    border-radius: 12px !important;
    border: 1px solid #d1d5db !important;
}

button {
    border-radius: 10px !important;
    height: 50px;
    font-size: 17px !important;
    font-weight: 600 !important;
}

footer {
    visibility: hidden;
}
"""


# Build Interface
with gr.Blocks(
    css=custom_css,
    theme=gr.themes.Soft(
        primary_hue="blue",
        secondary_hue="slate"
    )
) as demo:

    gr.HTML("""
    <div class="main-title">
        AI Language Translation Tool
    </div>

    <div class="sub-title">
        A modern Artificial Intelligence application that translates text instantly between multiple languages using Google Translator API.
    </div>
    """)

    with gr.Column(elem_classes="feature-box"):

        with gr.Row():

            source_lang = gr.Dropdown(
                choices=languages,
                value="english",
                label="Source Language"
            )

            target_lang = gr.Dropdown(
                choices=languages,
                value="urdu",
                label="Target Language"
            )

        text_input = gr.Textbox(
            lines=7,
            placeholder="Enter your text here...",
            label="Input Text"
        )

        translated_output = gr.Textbox(
            lines=7,
            label="Translated Text",
            show_copy_button=True
        )

        translate_button = gr.Button("Translate Text")

        translate_button.click(
            fn=translate_text,
            inputs=[text_input, source_lang, target_lang],
            outputs=translated_output
        )


# Launch Application
demo.launch()
```
